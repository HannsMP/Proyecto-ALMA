## Parte B — Primer contacto con conjuntos difusos (35 min)

Un **conjunto difuso** asigna un grado de pertenencia entre 0 y 1 a cada elemento del universo.  

Ejemplo: conjunto “**temperatura fría**” en un rango de 0 a 50 °C.


```
import numpy as np
import matplotlib.pyplot as plt

# Universo de discurso: temperatura entre 0 y 50 °C
x = np.linspace(0, 50, 500)

def trimf(x, a, b, c):
    y = np.zeros_like(x, dtype=float)
    idx = (a < x) & (x <= b)
    y[idx] = (x[idx] - a) / (b - a) if b != a else 0.0
    idx = (b < x) & (x < c)
    y[idx] = (c - x[idx]) / (c - b) if c != b else 0.0
    y[x == b] = 1.0
    return np.clip(y, 0, 1)

# Conjunto difuso: temperatura fría (0, 0, 20)
fria = trimf(x, 0, 0, 20)

plt.plot(x, fria, label="Temperatura fría")
plt.ylim(-0.05, 1.05)
plt.xlabel("Temperatura (°C)")
plt.ylabel("Grado de pertenencia")
plt.legend()
plt.show()
```

## Parte C — Operaciones básicas entre conjuntos difusos (35 min)

Podemos hacer **operaciones** entre conjuntos difusos:  
- Unión (máximo)  
- Intersección (mínimo)  
- Complemento (1 - µ)  


```
# Definimos otro conjunto: caliente (25,40,50)
caliente = trimf(x, 25, 40, 50)

# Unión (max)
union = np.maximum(fria, caliente)

# Intersección (min)
interseccion = np.minimum(fria, caliente)

# Complemento de "fría"
complemento_fria = 1 - fria

plt.figure()
plt.plot(x, fria, label="Fría")
plt.plot(x, caliente, label="Caliente")
plt.plot(x, union, label="Unión (fría ∪ caliente)")
plt.plot(x, interseccion, label="Intersección (fría ∩ caliente)")
plt.plot(x, complemento_fria, label="Complemento de fría")
plt.ylim(-0.05, 1.05)
plt.xlabel("Temperatura (°C)")
plt.ylabel("Grado de pertenencia")
plt.legend()
plt.show()
```

## Parte D — Mini reto (25 min)

Dado un valor puntual, por ejemplo **25 °C**, calculemos su pertenencia a los conjuntos `fría` y `caliente`.


```
valor = 25
mu_fria = np.interp(valor, x, fria)
mu_caliente = np.interp(valor, x, caliente)

print(f"A {valor} °C: pertenencia a fría = {mu_fria:.2f}, a caliente = {mu_caliente:.2f}")
```
